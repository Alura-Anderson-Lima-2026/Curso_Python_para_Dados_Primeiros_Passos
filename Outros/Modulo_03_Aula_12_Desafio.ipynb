3) Escreva um programa que determine se uma letra fornecida pela pessoa usuária é uma vogal ou consoante.

```python
v_letra = input('Informe uma letra: ')

v_letra = v_letra.upper()

if v_letra in 'A, E, I, O, U':
    print(f'A letra digitada "{v_letra}" é vogal')
else:
   print(f'A letra digitada "{v_letra}" é consoante')         
```

8) Escreva um programa que peça um número inteiro à pessoa usuária e determine se ele é par ou ímpar. Dica: Você pode utilizar o operador módulo %.

```python
v_numero = int( input('Informe um número inteiro: '))

if v_numero == 0:
    v_numero = ( v_numero % 0 )   
    print(f'O número digitado: {v_numero} é Par')
else:
    print(f'O número digitado: {v_numero} é Impar')      
```
