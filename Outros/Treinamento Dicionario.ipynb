7.2 DICIONARIOS


Os dicionários são uma estrutura de dados muito poderosa em Python, permitindo 
armazenar coleções de pares chave-valor. Ao contrário de listas ou tuplas, que 
armazenam elementos ordenados por índices, os dicionários associam uma chave 
única a um valor correspondente. Isso torna os dicionários ideais para armazenar 
informações que precisam ser facilmente acessadas e recuperadas através de 
chaves específicas.

7.2.1 Sintaxe e criação de um dicionário 


Os dicionários são criados utilizando chaves { } e cada par chave-valor é separado 
por dois pontos :. Diferentes pares de chave-valor são separados por vírgula, 
conforme a sintaxe a seguir: 

dicionario={chave1:valor1,chave2:valor2,...,chaveN:valorN} 


Assim como as listas e tuplas, dicionários podem conter qualquer tipo 
de dados. 

Podemos criar um dicionário onde cada par chave-valor é de um tipo diferente: 

```python
dicionario = {
    'nome': 'João',
    'Idade':39,
    1: 'Itu'
}

dicionario
```

```python
banco_de_dados = {
    'id_001':{
        'nome': 'Alice',
        'idade': 25,
        'cidade': 'São Paulo'
    },
    'id_002':{
        'nome':'Bob',
        'idade': 30,
        'cidade': 'Rio de Janeiro'
    },
    'id_003': {
        'nome':'Carol',
        'idade': 28,
        'cidade': 'Belo Horizonte'
    }
}

banco_de_dados
```

Também podemos criar um dicionário vazio. 


```python
dicionario_vazio = {}

dicionario_vazio
```

7.2.2 Acessando os elementos do dicionário 


Os valores em um dicionário podem ser acessados utilizando suas chaves, 
fornecendo o nome da chave entre colchetes [ ]. Da mesma forma, podemos 
modificar ou adicionar novos valores associados a uma chave, como mostra o 
exemplo a seguir.

```python
# Exemplo: acessando os elementos de um dicionário 

# Criando um dicionario

dic = {
    'nome': 'João',
    'idade': 30,
    'cidade': 'Itu'
}


```

```python
# Acessando um valor no dicionário
print(dic['nome'])
```

```python
# Modificando um valor no dicionário
dic['cidade'] = 35

dic
```

```python
# Adicionando um novo par chave-valor
dic['Profissão'] = 'Engenheiro'

dic

# Novos itens são adicionados a um dicionário quando fazemos uma atribuição a uma chave ainda não definida. 
```

Assim como nas listas e tuplas, podemos percorrer os elementos de um dicionário 
utilizando um comando de repetição, como o loop for. Podemos percorrer tanto o 
conjunto das chaves quanto os dos valores, como mostra o exemplo a seguir. 

Exemplo: iterando os elementos de um dicionário

```python
# Criar um dicionario
dic = {
    'nome': 'João',
    'Idade': 30,
    'cidade': 'Itu'
}

dic
```

```python
# Iterando nas chaves do dicionário
for chave in dic:
    print(chave)
```

```python
# Iterando nos valores do dicionário
for valor in dic.values():
    print(valor)
```

```python
# Iterando nos pares chave-valor do dicionário
for chave, valor in dic.items():
    print(f'{chave}:{valor}')
```
